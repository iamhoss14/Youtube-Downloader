```python
from pytubefix import YouTube

url = "https://www.youtube.com/watch?v=dQw4w9WgXcQ"  # public test video
yt = YouTube(url)

print(yt.title)

stream = (
    yt.streams
    .filter(file_extension="mp4", type="video")
    .order_by("resolution")
    .desc()
    .first()
)

if stream is None:
    raise RuntimeError("No downloadable MP4 stream found for this video")

stream.download(output_path="downloads")
```

```python
from pytubefix import YouTube
from pathlib import Path

url = "https://www.youtube.com/watch?v=dQw4w9WgXcQ"  # public video
yt = YouTube(url)

print("Title:", yt.title)

stream = (
    yt.streams
    .filter(file_extension="mp4", type="video")
    .order_by("resolution")
    .desc()
    .first()
)

if stream is None:
    raise RuntimeError("No MP4 video stream found for this URL")

out_dir = Path("downloads")
out_dir.mkdir(exist_ok=True)
file_path = stream.download(output_path=str(out_dir))
print("Downloaded:", file_path)
```

```python
import yt_dlp

url = "https://www.youtube.com/watch?v=EK0hksG9VHI"

try:
    # Configure yt-dlp with proper options to bypass bot detection
    ydl_opts = {
        'quiet': False,
        'no_warnings': False,
        'socket_timeout': 30,
    }
    
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        info = ydl.extract_info(url, download=False)
    
    print(f"Title: {info.get('title', 'N/A')}")
    print(f"Duration: {info.get('duration', 'N/A')} seconds")
    
except yt_dlp.utils.DownloadError as e:
    print(f"Download Error: {str(e)}")
    print("\nNote: YouTube may require:")
    print("1. Browser cookies (--cookies-from-browser)")
    print("2. A JavaScript runtime (Node.js/Deno)")
    print("3. Authentication with valid credentials")
    
except Exception as e:
    print(f"Error: {type(e).__name__}: {str(e)}")
```
