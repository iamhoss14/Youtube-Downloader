```python
from pytubefix import YouTube

url = "https://www.youtube.com/watch?v=dQw4w9WgXcQ"  # public test video
yt = YouTube(url)

print(yt.title)

stream = (
    yt.streams
    .filter(file_extension="mp4", type="video")
    .order_by("resolution")
    .desc()
    .first()
)

if stream is None:
    raise RuntimeError("No downloadable MP4 stream found for this video")

stream.download(output_path="downloads")
```

```python
from pytubefix import YouTube
from pathlib import Path

url = "https://www.youtube.com/watch?v=dQw4w9WgXcQ"  # public video
yt = YouTube(url)

print("Title:", yt.title)

stream = (
    yt.streams
    .filter(file_extension="mp4", type="video")
    .order_by("resolution")
    .desc()
    .first()
)

if stream is None:
    raise RuntimeError("No MP4 video stream found for this URL")

out_dir = Path("downloads")
out_dir.mkdir(exist_ok=True)
file_path = stream.download(output_path=str(out_dir))
print("Downloaded:", file_path)
```

```python
import yt_dlp


def get_video_info(url: str):
  """Extracts and prints metadata about a YouTube video without downloading it."""
  ydl_opts = {
      "extract_flat": False,  # Get full info including stream formats
  }

  try:
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
      print(f"Extracting metadata for: {url}\n")

      # Extract info without triggering a download
      info = ydl.extract_info(url, download=False)

      # Pull key details from the dictionary
      title = info.get("title", "Unknown Title")
      uploader = info.get("uploader", "Unknown Uploader")
      duration = info.get("duration", 0)  # in seconds
      views = info.get("view_count", 0)

      # Format duration nicely
      minutes, seconds = divmod(duration, 60)

      print(f"📌 Title: {title}")
      print(f"👤 Uploader: {uploader}")
      print(f"⏱️ Duration: {minutes}m {seconds}s")
      print(f"👁️ Views: {views:,}")
      print("-" * 40)

      # Inspect a few available formats/resolutions
      print("🎞️ Sample Available Formats:")
      formats = info.get("formats", [])
      for f in formats[:6]:  # Show a small sample of formats
        print(
            f" - ID: {f.get('format_id')} | Res:"
            f" {f.get('resolution', 'audio only')} | Ext: {f.get('ext')}"
        )

  except Exception as e:
    print(f"An error occurred: {e}")


if __name__ == "__main__":
  test_url = "https://www.youtube.com/watch?v=dQw4w9WgXcQ"
  get_video_info(test_url)
```

```python
import os
import yt_dlp


def print_video_title(url: str, cookie_file: str = "cookies.txt"):
  ydl_opts = {
      # extract_flat speeds up the process since we only need basic metadata
      "extract_flat": True,
  }

  # Automatically use cookies.txt if it exists in the folder
  if os.path.exists(cookie_file):
    ydl_opts["cookiefile"] = cookie_file
    print("🍪 Using cookies.txt for authentication.")

  try:
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
      info = ydl.extract_info(url, download=False)
      title = info.get("title", "Unknown Title")
      print(f"\n📌 Video Title: {title}\n")
  except Exception as e:
    print(f"❌ An error occurred: {e}")


if __name__ == "__main__":
  test_url = input("Enter YouTube URL: ").strip()
  if test_url:
    print_video_title(test_url)
```
